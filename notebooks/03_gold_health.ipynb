# Gold Layer - Aggregated Dashboard Metrics

This notebook reads from `health_catalog.silver.*`, computes final aggregates, and writes to `health_catalog.gold.*`.

`````python
from pyspark.sql import SparkSession
from databricks_health import GoldAggregationOrchestrator

spark = SparkSession.builder.getOrCreate()

def main():
    orchestrator = GoldAggregationOrchestrator(spark, catalog='health_catalog')
    orchestrator.run()

if __name__ == '__main__':
    main()
`````

## Dashboard Query Examples

### Infrastructure health tab
````sql
SELECT * FROM health_catalog.gold.infra_metrics
````

### Pipeline health tab
````sql
SELECT job_id, name, total_runs, successful_runs, failed_runs, avg_duration_minutes
FROM health_catalog.gold.pipeline_metrics
ORDER BY failed_runs DESC
````

### Data product health tab
````sql
SELECT catalog_name, table_count, stale_table_count
FROM health_catalog.gold.data_product_metrics
````

### Query performance metrics
````sql
SELECT avg_query_ms, slow_query_count, query_count
FROM health_catalog.gold.query_performance_metrics
````
