# Bronze Layer - Raw Health Data Ingestion

This notebook ingests raw Databricks platform telemetry from REST APIs, system tables, and Azure Monitor into bronze Delta tables under Unity Catalog.

Target: `health_catalog.bronze.<table>`

Sources:
- Databricks REST APIs: clusters, jobs, runs, pipelines, DLT, SQL warehouses, Apps.
- Databricks system tables and catalog views: query history, job history, cluster metadata, table metadata.
- Azure Monitor / AzureDiagnostics logs for resource health, storage connectivity, and workspace uptime.

```python
import os
from databricks_health.ingestors import BronzeIngestionOrchestrator

DATABRICKS_HOST = os.getenv('DATABRICKS_HOST')
DATABRICKS_TOKEN = os.getenv('DATABRICKS_TOKEN')
AZURE_LOG_ANALYTICS_WORKSPACE_ID = os.getenv('AZURE_LOG_ANALYTICS_WORKSPACE_ID')

def main():
    orchestrator = BronzeIngestionOrchestrator(
        databricks_host=DATABRICKS_HOST,
        databricks_token=DATABRICKS_TOKEN,
        azure_log_analytics_workspace_id=AZURE_LOG_ANALYTICS_WORKSPACE_ID,
        catalog='health_catalog',
        schema='bronze',
        lookback_days=30,
    )
    orchestrator.run()

if __name__ == '__main__':
    main()
```

## Bronze Tables (health_catalog.bronze.*)

| Table | Source | Description |
|-------|--------|-------------|
| clusters | REST API | Cluster metadata and state |
| job_runs | REST API | Job run history |
| pipelines | REST API | Job/pipeline definitions |
| dlt_pipelines | REST API | DLT pipeline definitions |
| sql_warehouses | REST API | SQL warehouse metadata |
| sql_query_history | system.query.history | SQL query execution history |
| job_history | system.job.history | Job execution history |
| cluster_metadata | system.compute.clusters | Detailed cluster metadata |
| table_metadata | system.information_schema.tables | Unity Catalog table inventory |
| azure_monitor_clusters | AzureDiagnostics | Cluster resource health metrics |
| azure_monitor_storage | AzureDiagnostics | Storage connectivity metrics |
| azure_monitor_workspace | AzureDiagnostics | Workspace uptime metrics |
